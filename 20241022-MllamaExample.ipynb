```python
model = transformers.MllamaForConditionalGeneration.from_pretrained(
    "meta-llama/Llama-3.2-11B-Vision",
    torch_dtype="float32",
)
processor = transformers.AutoProcessor.from_pretrained(model.config._name_or_path)
```

### Inputs


We ignore the batch dimension.

Params:

| Param | Value |
| --- | --- |
| `max_sequence_length` | 131,072 |
| `vocab_size` | 128,256 |
| `image_size` | 448 |
| `max_n_tiles` | 4 |
| `img_mean` | `[0.48145466, 0.4578275, 0.40821073]` |
| `img_std` | `[0.26862954, 0.26130258, 0.27577711]` |

Shapes:

 - `input_ids` of shape `(sequence_length)` in `[0, vocab_size)`
   - Format: `['<|begin_of_text|>', '<|image|>', '<|begin_of_text|>', ...]`
 - `attention_mask` of shape `(sequence_length)`
   - `1` for unmasked tokens
 - `cross_attention_mask` of shape `(sequence_length, n_images, n_tiles)`
   - `1` for unmasked token-tile
   - `0` for the first token `<|begin_of_text|>`
   - `0` for absent tiles
 - `pixel_values` of shape `(n_images, n_tiles, 3, image_size, image_size)`
   - Tiles in scan order e.g. `[top-left, top-right, bottom-left, bottom-right]`
   - Colour channels `[R, G, B]`
   - Pixels in scan order `img[-y, x]`
   - Right & bottom -padded with `0`
   - Small images are stretched so that their larger dimension fills `image_size` & the other is padded
   - Large images are squashed to that their larger dimension fills `2 * image_size` & the other is padded
   - Normalised image: `((raw_data / 255) - img_mean) / img_std` (mean and std from [preprocessor_config.json](https://huggingface.co/meta-llama/Llama-3.2-11B-Vision/blob/main/preprocessor_config.json))
 - `aspect_ratio_id` of shape `(n_images)`
   - Index into `config.vision_config.supported_aspect_ratios` e.g. `6 => [2, 2]` (2 tiles by 2 tiles), and `0 => [1, 1]`
 - `aspect_ratio_mask` of shape `(n_images, n_tiles)`
   - `1` for unmasked tiles

# Implementation

```python
with torch.no_grad(), tensor_tracker.track(model, include="^vision_model.layernorm_pre$") as tracker:
    out = model(**inputs)
    print(processor.tokenizer.convert_ids_to_tokens(out.logits[0, -1:].argmax(-1)))
print(tracker[0].value.shape)
```

```python
p = {k.replace("vision_model.", ""): v for k, v in model.named_parameters() if k.startswith("vision_model.")}

patch_size = p["patch_embedding.weight"].shape[-1]
n_patches = (inputs.pixel_values.shape[-1] // patch_size)**2
n_tiles = inputs.pixel_values.shape[-4]
n_aspect_ratios = p["pre_tile_positional_embedding.embedding.weight"].shape[0]
hidden_size = p["patch_embedding.weight"].shape[0]
norm_eps = 1e-5

input_pixels = inputs.pixel_values.flatten(end_dim=2)
aspect_ratio_id = inputs.aspect_ratio_ids.reshape(())

def layer_norm(x: Tensor, weight: Tensor, bias: Tensor) -> Tensor:
    z = x - x.mean(-1, keepdim=True)
    z /= torch.sqrt((z ** 2).mean(-1, keepdim=True) + norm_eps)
    return z * weight + bias

# Patch encoding & positional encoding
gated_embeddings = (
    p["gated_positional_embedding.embedding"]
    * (1 - p["gated_positional_embedding.gate"].tanh().reshape(()))
    + p["gated_positional_embedding.tile_embedding.weight"].reshape(n_aspect_ratios, n_tiles, n_patches + 1, hidden_size)
    * p["gated_positional_embedding.gate"].tanh().reshape(())
)
positional_embedding_combined = (
    (p["pre_tile_positional_embedding.embedding.weight"].reshape(n_aspect_ratios, n_tiles, 1, hidden_size)
     * p["pre_tile_positional_embedding.gate"].tanh())
    + gated_embeddings[..., 1:, :]
)
class_embedding_combined = p["class_embedding"] + gated_embeddings[..., 0, :]
patches = F.conv2d(
    input_pixels, p["patch_embedding.weight"], stride=patch_size,
).flatten(start_dim=-2).transpose(-1, -2)
patches += positional_embedding_combined[aspect_ratio_id]
patches = torch.cat([class_embedding_combined[aspect_ratio_id].unsqueeze(-2), patches], dim=-2)
patches = layer_norm(patches, p["layernorm_pre.weight"], p["layernorm_pre.bias"])

# Activations

# Weights

# patches.shape
torch.testing.assert_close(patches, tracker[0].value.squeeze(0))
```
