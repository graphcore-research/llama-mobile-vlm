```python
from pathlib import Path
import json
from eval import vqa
```

```python
path = Path("out/vqa/2025-10-21T17-42-33")
```

```python
all_results = {}
for task_name in ["ai2d", "chartqa", "docvqa", "vqa"]:
    print(f"TASK: {task_name}")
    results = []
    with open(path / f"{task_name}.jsonl") as f:
        for line in f:
            results.append(json.loads(line))
    all_results[task_name] = results
    metrics = vqa.TASK_METRIC[task_name]
    for m in metrics:
        out = [x[m] for x in results]
        avg = sum(out) / len(out)
        print(f"{m}: {avg}")
    print()
```
