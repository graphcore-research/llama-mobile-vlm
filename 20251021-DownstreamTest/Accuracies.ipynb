```python
import json
import math
from pathlib import Path

from eval import vqa
```

```python
path = Path("out/vqa/2025-10-21T17-49-27")
```

```python
meta_acc = {"ai2d": 0.911, "chartqa": 0.834, "docvqa": 0.884, "vqa": 0.752}

all_results = {}
for task_name in ["ai2d", "chartqa", "docvqa", "vqa"]:
    print(f"TASK: {task_name} [Meta: {meta_acc[task_name]}]")
    results = []
    with open(path / f"{task_name}.jsonl") as f:
        for line in f:
            results.append(json.loads(line))
    all_results[task_name] = results
    metrics = vqa.TASK_METRIC[task_name]
    for m in metrics:
        out = [x[m] for x in results]
        avg = sum(out) / len(out)
        se = math.sqrt(avg * (1 - avg) / len(out))
        print(f"{m}: {avg} ± {2*se}")
    print()
```
