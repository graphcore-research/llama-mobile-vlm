```python
import json
from pathlib import Path
import transformers
```

```python
path = Path("data/imagenet-train-generation/c146da/out/out0.jsonl")
```

```python
examples = []
with open(path) as f:
    for line in f:
        idx, text = json.loads(line)
        examples.append((idx, text))
        
```

```python
tokenizer = transformers.AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-11B-Vision-Instruct")
```

```python
len(tokenizer(examples[0][1])["input_ids"])
```

```python
lens = [len(tokenizer(x[1])["input_ids"]) for x in examples]
```

```python
import matplotlib.pyplot as plt
```

```python
_ = plt.hist(lens, bins=20, color='skyblue', edgecolor='black')
```
